## `04`
# $$Using\space  SQL$$


`i` Create Temp View:

```python
df.createOrReplaceTempView("org")
```

`ii` Display `"RAM"`column and `count` "RAM" column:

```python
spark.sql("SELECT RAM, count(RAM) FROM org GROUP BY RAM").show()
```

`iii` Get all columns when `"Year"` column equal `"2015"`  

```python
spark.sql("SELECT * FROM global_temp.temp WHERE Year = 2015").show()
```

`iv` Get all when `"Model"` start with `"M"`

```python
spark.sql("SELECT * FROM org WHERE Model LIKE 'M%'").show()
```

`v` Get all data when `"Model"` column equal `"MacBook Pro"`

```python
spark.sql("SELECT * FROM org WHERE Model = 'MacBook Pro'").show()
```

`vi` Get all data with `Multiple Conditions` when `"RAM"` column equal `"8GB"` and `"Model"` column is `"Macbook"`.

```python
spark.sql("SELECT * FROM org WHERE RAM = '8GB' AND Model = 'MacBook'").show()
```

`vii` Get all data with `Multiple Conditions` when `"D"` greater than or equal `"8"` and `"Model"` column is `"iMac"`.

```python
spark.sql("SELECT * FROM org WHERE D >= 8 AND Model = 'iMac'").show()
```
