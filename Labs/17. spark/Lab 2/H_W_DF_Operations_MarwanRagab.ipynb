```python
from pyspark.sql.functions import format_number
```

```python
result = df.describe()
result.select(result['summary'],
              format_number(result['Open'].cast('float'),2).alias('Open'),
              format_number(result['High'].cast('float'),2).alias('High'),
              format_number(result['Low'].cast('float'),2).alias('Low'),
              format_number(result['Close'].cast('float'),2).alias('Close'),
              format_number(result['Volume'].cast('float'),2).alias('Volume')).show()
```

`###` Create a new dataframe with a column called `HV Ratio` that is the ratio of the `High Price versus volume of stock traded for a day`.

```python
df_new = df.withColumn('HV Ratio', df['High']/df['Volume'])
df_new.select('HV Ratio').show()
```

`#` **What day had the `Peak High` in `Price`?**

```python
df_new.orderBy(df_new['High'].desc()).head(1)[0][0]
```

`##` **What is the `mean` of the `Close` column?**

```python
df_new.agg({'Close':'mean'}).show()
```

`###` **What is the `max` and `min` of the `Volume` column?**

```python
df_new.agg(max('Volume'), min('Volume')).show()
```

`####` **How many days was the Close `lower than 60` dollars?**

```python
df_new.filter(df_new['Close'] < 60).count()
```

`#####` **What `percentage of the time` was the `High greater than 80`dollars ?**
-  In other words, (Number of Days High>80)/(Total Days in the dataset)

```python
df_new.filter(df_new['High'] > 80).count()/df_new.count()*100
```

`#####` **What is the `Pearson correlation` between `High and Volume?`**

```python
df_new.corr('High', 'Volume')
```

`######` **What is the `max High` per year?**

```python
df_arr = df_new.withColumn('Year', year(df_new['Date']))
df_arr.groupBy('Year').max('High').show()
```

`####` What is the `average Close` for each `Calendar Month`?
  - In other words, across all the years, what is the average Close price for `Jan,Feb, Mar, etc...` Your result will have a value for each of these months.

```python
df_arr = df_new.withColumn('Month', month(df_new['Date']))
df_arr.groupBy('Month').mean('Close').orderBy('Month').show()
```
