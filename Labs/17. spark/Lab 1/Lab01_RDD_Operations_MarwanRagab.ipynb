### Read sample1.txt file into RDD and displaying the first 4 elements

```python
text_data = '''
Utilitatis causa amicitia est quaesita.
Lorem ipsum dolor sit amet, consectetur adipiscing elit.
Collatio igitur ista tenihil iuvat.
Honesta oratio, Socratica, Platonis etiam.
Primum in nostranepotestate est, quid meminerimus?
Duo Reges: constructio interrete.
Quid, sietiam iucunda memoria est praeteritorum malorum?
Si quidem, inquit, tollerem,'''

lines_list = text_data.strip().splitlines()
lines_rdd = sc.parallelize(lines_list)
lines_rdd.saveAsTextFile("my_output_folder1")
lines_rdd.collect()
```

### Count the total number of rows in RDD

```python
lines_rdd.count()
```

### Create a function to convert the data into lower case and splitting it

```python
filter_rdd1 = lines_rdd.map(lambda x: x.lower())
filter_rdd2 = filter_rdd1.map(lambda x: x.split())
filter_rdd2.collect()
```

### Remove the stopwords from the previous text. i.e. Remove it.

```python
stopwords = ['a','all','the','as','is','am','an','and',
             'be','been','from','had','I','I’d','why','with']
# Hint: you may need use flatMap
```

```python
filter_rdd3 = filter_rdd2.flatMap(lambda x:[word for word in x if word not in stopwords])
filter_rdd3.collect()
```

### Find the words starting with ‘c’

```python
filter_rdd4 = filter_rdd3.filter(lambda x :x.startswith('c'))
filter_rdd4.collect()
```

### Reduce the data by key and sum it (use the data from the following list)

```python
list = [('JK', 22), ('V', 24), ('Jimin',24), ('RM', 25)
        , ('J-Hope', 25), ('Suga', 26), ('Jin', 27)
       , ('J-Hope', 12), ('Suga', 25), ('Jin', 34)
       , ('JK', 32), ('V', 44), ('Jimin',14), ('RM', 35)]
# Hint: use reduceByKey
```

```python
filter2 = sc.parallelize(list)
filter2.collect()
```

```python
filter_rdd3 = filter2.reduceByKey(lambda x,y:x+y)
filter_rdd3.collect()
```
