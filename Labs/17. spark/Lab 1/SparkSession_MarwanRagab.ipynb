```python
rddfile_split_flat = rddfile.flatMap(lambda x: x.split(','))
```

```python
rddfile_split_flat.take(10)
```

```python
rddfile_split_flat1 = rddfile_split_flat.flatMap(lambda x: x.split(" "))\
.filter(lambda x: len(x) > 0 and x.strip() != '')
```

```python
rddfile_split_flat1.take(10)
```

```python
rddfile_split_flat2 = rddfile_split_flat1.keyBy(lambda x : x[0])
```

```python
rddfile_split_flat2.take(10)
```

```python
rddfile_split_flat2.take(1000)
```

```python
rddfile_split_flat3 = rddfile_split_flat2.groupByKey()
rddfile_split_flat3
```

```python
rddfile_split_flat3.take(10)
```

```python
for k ,v in rddfile_split_flat3.take(10):
  print(k,list(v))
```
