### for each issue adapt this methodology

**First issue**

```python
#solution
#first issue is object values
#detect coulmn is object
print("Original object columns:")
object_cols = []
for col in df_copy.columns:
    if df_copy[col].dtype == object:
        print(col)
        object_cols.append(col)


non_numeric_object_cols = ['Country', 'Region']


numeric_like_object_cols = [col for col in object_cols if col not in non_numeric_object_cols]

print("\nColumns identified for numeric conversion:")
for col in numeric_like_object_cols:
    print(col)

# Apply cleaning and conversion to numeric
for col in numeric_like_object_cols:
    df_copy[col] = df_copy[col].astype(str)

    df_copy[col] = df_copy[col].str.replace(',', '.', regex=False)
    df_copy[col] = df_copy[col].str.replace('$', '', regex=False)
    df_copy[col] = df_copy[col].str.replace('%', '', regex=False)
    df_copy[col] = df_copy[col].str.replace(' ', '', regex=False)
    df_copy[col] = df_copy[col].str.replace('(', '', regex=False)
    df_copy[col] = df_copy[col].str.replace(')', '', regex=False)
    df_copy[col] = df_copy[col].str.replace(':', '', regex=False)
    df_copy[col] = df_copy[col].str.replace('-', '', regex=False)


    df_copy[col] = pd.to_numeric(df_copy[col], errors='coerce')
```

```python
#test
df_copy.info()
```

**Second issue**

```python
#solution
df_copy.isnull().sum()
null_cols_list=[]
for col in df_copy.columns:
  if df_copy[col].isnull().any():
    null_cols_list.append(col)


for col in null_cols_list:
  if df_copy[col].dtype == object:
    df_copy[col].fillna(df_copy[col].mode()[0], inplace=True)
  else:
    df_copy[col].fillna(df_copy[col].median(), inplace=True)
```

```python
#test
df_copy.isnull().sum()
```

```python
df_copy.head()
```

**Third issue**

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
df_copy['Region']= label_encoder.fit_transform(df_copy['Region'])

df_copy = pd.get_dummies(df_copy, columns=['Country'])
```

```python
for col in df_copy.columns:
  plt.figure(figsize=(15, 10))
  sns.histplot(df_copy[col], kde=True,color='skyblue')
  skwee=df_copy[col].skew()
  plt.title(f'Distribution of {col} with Skewness: {skwee:.2f}')
  plt.xlabel(col)
  plt.ylabel('Frequency')
  plt.show()


```
