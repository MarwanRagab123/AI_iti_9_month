```python
from selenium import webdriver
driver = webdriver.Edge()
```

```python
URL = "https://www.amazon.in/gp/bestsellers/books/"
driver.get(URL)
 
driver.implicitly_wait(5)
 
```

```python
prices=driver.find_elements("xpath","//*/span[@class='a-size-base a-color-price']/span[@class='_cDEzb_p13n-sc-price_3mJ9Z']")
```

```python
print(len(prices))
```

```python
for pric in prices:
    print(pric.text)
```
