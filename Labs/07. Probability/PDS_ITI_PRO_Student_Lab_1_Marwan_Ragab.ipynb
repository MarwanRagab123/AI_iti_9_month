# Problem 1

Use numpy to simulate the vector outcomes for repeating the following random experiments $1000$ times and use the relative frequency to compute the probability of the corresponding events:
    
1. **Experiment:** Flipping a fair coin, **Event:** A head occurs.
     
Observe how changing the number of repetition affects the computed probabilities.

**Hints:**
- Use [numpy.random.randint](https://numpy.org/doc/stable/reference/random/generated/numpy.random.randint.html) or [numpy.random.choice](https://numpy.org/doc/stable/reference/random/generated/numpy.random.choice.html) to simulate the experiment.
- Other useful functions [numpy.arange](https://numpy.org/doc/stable/reference/generated/numpy.arange.html), [numpy.cumsum](https://numpy.org/doc/stable/reference/generated/numpy.cumsum.html), [numpy.sum](https://numpy.org/doc/stable/reference/generated/numpy.sum.html), and [matplotlib.pyplot.plot](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html)

```python
#type your code here:
for n in [10,50,100,500,1000,5000,10000,50000,100000]:
  a=np.random.choice([0,1],size=n)
  b=np.sum(a)/n
  print(f"n = {n} ,P(e) = {b}")
  print("-"*50)

cum=np.cumsum(a)
k=np.arange(1,n+1)
x=cum/k
plt.plot(k,x)
plt.axhline(0.5,color='r')
plt.grid()
```

2. **Experiment:** Rolling a fair die, **Event:** Number 5 occurs.

```python
#type your code here:
for n in [10,50,100,500,1000,5000,10000,50000,100000]:
  a=np.random.choice([0,1,2,3,5,6],size=n)
  b=np.sum(a==5)/n
  print(f"n = {n} ,P(e) = {b}")
  print("-"*50)

cum=np.cumsum(np.random.choice([0,1,2,3,5,6],size=1000)==5)
k=np.arange(1,1001)
x=cum/k
plt.plot(k,x)
plt.axhline(0.16,color='r')
plt.grid()
```

3. **Experiment:** Rolling a fair die, **Event:** A number greater than 3 occurs.

```python
#type your code here:
for n in [10,50,100,500,1000,5000,10000,50000,100000]:
  a=np.random.choice([0,1,2,3,5,6],size=n)
  b=np.sum(a>3)/n
  print(f"n = {n} ,P(e) = {b}")
  print("-"*50)

cum=np.cumsum(np.random.choice([0,1,2,3,5,6],size=1000)>3)
k=np.arange(1,1001)
x=cum/k
plt.plot(k,x)
plt.grid()
```

4. **Experiment:** Rolling two fair dice, **Event:** Sum of two dice equal 5.

```python
#type your code here:

for n_trials in [10,50,100,500,1000,5000,10000,50000,100000]:

  a = np.random.choice([1, 2, 3, 4, 5, 6] , size=(n_trials, 2))
  sums_sim = a.sum(axis=1)
  count_sum_5 = np.sum(sums_sim == 5)

  prob_estimate = count_sum_5 / n_trials
  print(f"n = {n_trials} ,P(e) = {prob_estimate}")
  print("-"*50)



rolls_for_plot = np.random.choice([1, 2, 3, 4, 5, 6] , size=(n_trials, 2))
sums = rolls_for_plot.sum(axis=1)
event_occurred_for_plot = (sums == 5)
cum_event_occurrences = np.cumsum(event_occurred_for_plot)
k_plot = np.arange(1, n_trials + 1)
x_plot = cum_event_occurrences / k_plot
plt.figure(figsize=(10,6))
plt.plot(k_plot, x_plot)
plt.axhline(4/36, color='r')
plt.show()
```

# Problem 3: Independence

Use NumPy to create simulations and compute **empirical  probability** for the following outcomes. Compare the results with the **theoretical probability**.

1. Two fair coin flips produce exactly two heads

```python
#type your code here:

# simulate 1 million tests of two fair coin flips
num_tests = 1000000
flips=np.random.choice([0, 1], size=(num_tests, 2)) # 0 for tails 1 for heads
# sums of all tests
sums=flips.sum(axis=1)
two_heads=np.sum(sums == 2)


# proportion of tests that produced exactly one head
proportion=two_heads / num_tests
print(f"{proportion}")
```

2. Three fair coin flips produce exactly one head

```python
#type your code here:

# simulate 1 million tests of three fair coin flips
num_tests = 1000000
flips = np.random.choice([0, 1], size=(num_tests, 3))
# sums of all tests
sums=flips.sum(axis=1)
one_heads=np.sum(sums==1)

# proportion of tests that produced exactly one head
proportion = one_heads / num_tests
print(f"{proportion}")
```

3. Three biased coin flips with P(H) = 0.6 produce exactly one head

```python
#type your code here:


# simulate 1 million tests of three biased coin flips
# hint: use np.random.choice()
num_tests = 1000000
flips = np.random.choice([0, 1], size=(num_tests, 3),p=[0.4,0.6])

# sums of all tests
sums=flips.sum(axis=1)
one_heads=np.sum(sums==1)

# proportion of tests that produced exactly one head
proportion = one_heads / num_tests
print(f"{proportion}")
```

# Problem 5: Correlation and Covariance

The CSV file includes a hourly/daily summary for Szeged, Hungary area, between 2006 and 2016. After inspecting the data:

- Calculate the covariance and correlation coeffecient between temperature and apparent temperature
- Calculate the covariance and correlation coeffecient between temperature and humidity

```python
import zipfile
import os
import pandas as pd
zip_file_path = '/content/drive/MyDrive/weatherHistory.csv.zip'
extraction_path = '/content/drive/MyDrive/'

with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(extraction_path)

print(f"File extracted to: {extraction_path}")
```

```python
df = pd.read_csv('/content/drive/MyDrive/weatherHistory.csv')
temperature = df['Temperature (C)'].to_numpy().reshape(-1,1)
apparant_temp = df['Apparent Temperature (C)'].to_numpy().reshape(-1,1)
humidity = df['Humidity'].to_numpy().reshape(-1,1)
```

## Temperature and Apparent Temperature

```python
#type your code here:
cov = np.cov(temperature.flatten(), apparant_temp.flatten())[0, 1]
print(f"Covariance: \n{cov}\n")
cov2 = np.cov(temperature.flatten(), apparant_temp.flatten())
print(f"np.cov: \n{cov2}\n")
corr = np.corrcoef(temperature.flatten(), apparant_temp.flatten())[0, 1]

print(f"Correlation Cof: \n{corr}\n")
corr2 = np.corrcoef(temperature.flatten(), apparant_temp.flatten())

print(f"np.corrcoef: \n{corr2}\n")
```
